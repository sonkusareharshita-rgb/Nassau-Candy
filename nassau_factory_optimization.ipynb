```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True, errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True, errors='coerce')

df['lead time'] = (df['Ship Date'] - df['Order Date']).dt.days

print(df['lead time'].isna().sum())
print(df['lead time'].describe())
```

```python
X = df[['Product Name', 'Region', 'Ship Mode']]
y = df['lead time']

print(X.shape)
print(y.shape)
```

```python
X_encoded = preprocessor.fit_transform(X)

print("Encoding complete!")
print("Shape:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print("Training:", X_train.shape[0])
print("Testing:", X_test.shape[0])
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)

print("Linear Regression complete!")
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test, lr_pred))
mae = mean_absolute_error(y_test, lr_pred)
r2 = r2_score(y_test, lr_pred)

print("RMSE:", rmse)
print("MAE:", mae)
print("R²:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

print("Random Forest complete!")
```

```python
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_r2 = r2_score(y_test, rf_pred)

print("RMSE:", rf_rmse)
print("MAE:", rf_mae)
print("R²:", rf_r2)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = GradientBoostingRegressor(random_state=42)

gb_model.fit(X_train, y_train)

gb_pred = gb_model.predict(X_test)

print("Gradient Boosting complete!")
```

```python
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_r2 = r2_score(y_test, gb_pred)

print("RMSE:", gb_rmse)
print("MAE:", gb_mae)
print("R²:", gb_r2)
```

```python
factory_map = {
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar -Scrumdiddlyumptious": "Lot's O' Nuts",
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",
    "Everlasting Gobstopper": "Secret Factory",
    "Hair Toffee": "The Other Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",
    "Kazookles": "The Other Factory"
}

df["Current Factory"] = df["Product Name"].map(factory_map)

print("Missing Factory:", df["Current Factory"].isna().sum())
```

```python
X = df[['Product Name', 'Current Factory', 'Region', 'Ship Mode']]
y = df['lead time']

categorical_features = ['Product Name', 'Current Factory', 'Region', 'Ship Mode']

preprocessor_factory = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

X_encoded_factory = preprocessor_factory.fit_transform(X)

print("Factory feature added!")
print(X_encoded_factory.shape)
```

```python
X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(
    X_encoded_factory, y, test_size=0.2, random_state=42
)

rf_factory = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_factory.fit(X_train_f, y_train_f)

rf_factory_pred = rf_factory.predict(X_test_f)

print("Factory-aware model complete!")
```

```python
factories = [
    "Lot's O' Nuts",
    "Wicked Choccy's",
    "Sugar Shack",
    "Secret Factory",
    "The Other Factory"
]

scenario_rows = []

for product in df["Product Name"].unique():
    product_data = df[df["Product Name"] == product].iloc[0]

    for factory in factories:
        scenario_rows.append({
            "Product Name": product,
            "Current Factory": product_data["Current Factory"],
            "Scenario Factory": factory,
            "Region": product_data["Region"],
            "Ship Mode": product_data["Ship Mode"]
        })

scenario_df = pd.DataFrame(scenario_rows)

print("Scenarios created:", len(scenario_df))
scenario_df.head()
```

```python
X_scenario = scenario_df[
    ['Product Name', 'Scenario Factory', 'Region', 'Ship Mode']
].copy()

X_scenario = X_scenario.rename(
    columns={'Scenario Factory': 'Current Factory'}
)

X_scenario_encoded = preprocessor_factory.transform(X_scenario)

scenario_df["Predicted Lead Time"] = rf_factory.predict(X_scenario_encoded)

scenario_df.head()
```

```python
X_scenario = scenario_df[
    ['Product Name', 'Scenario Factory', 'Region', 'Ship Mode']
].copy()

X_scenario = X_scenario.rename(
    columns={'Scenario Factory': 'Current Factory'}
)

X_scenario_encoded = preprocessor_factory.transform(X_scenario)

scenario_df["Predicted Lead Time"] = rf_factory.predict(X_scenario_encoded)

scenario_df.head()
```

```python
current_lead = (
    df.groupby("Product Name")["lead time"]
    .mean()
    .reset_index()
    .rename(columns={"lead time": "Current Lead Time"})
)

recommendations = scenario_df.merge(
    current_lead,
    on="Product Name",
    how="left"
)

recommendations["Lead Time Reduction"] = (
    recommendations["Current Lead Time"]
    - recommendations["Predicted Lead Time"]
)

recommendations["Lead Time Reduction %"] = (
    recommendations["Lead Time Reduction"]
    / recommendations["Current Lead Time"]
) * 100

best_factory = (
    recommendations
    .sort_values(["Product Name", "Predicted Lead Time"])
    .groupby("Product Name")
    .first()
    .reset_index()
)

best_factory[
    ["Product Name", "Current Factory", "Scenario Factory",
     "Current Lead Time", "Predicted Lead Time",
     "Lead Time Reduction %"]
]
```
